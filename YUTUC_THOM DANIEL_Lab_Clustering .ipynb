## Part 2 — K-Means Clustering

***Use the elbow method and silhouette analysis to choose K. Plot both curves and justify your final choice.***

```python
# Elbow Method
K_range = range(1, 10)
wcss = []

print("Running K-Means for K = 1 to 10...")
print()

for k in K_range:
    km = KMeans(n_clusters=k, init='k-means++', n_init=10, random_state=42)
    km.fit(df_scaled)
    wcss.append(km.inertia_)
    print(f"  K={k:2d}  |  WCSS = {km.inertia_:8.2f}")

# Plot
plt.figure(figsize=(8,5))
plt.plot(K_range, wcss, marker='o', color = 'steelblue', linewidth = 2.5, 
        markersize = 8, markerfacecolor = 'coral', markeredgecolor = 'white', markeredgewidth = 1.5)
plt.xlabel("Number of Clusters (K)")
plt.ylabel("WCSS")
plt.title("Elbow Method")
plt.xticks(list(K_range))
plt.axvline(x=2, color='red', linestyle='--', alpha=0.6, label='Elbow at K=2')
plt.legend()
plt.show()
```

**Insight:**
- Looking at the Graph it can be seen that the largest drop of K happened in K = 1 going to K = 2. It shows that the elbow curve made a sharp bend around K = 2.

```python
# Silhouette Analysis

K_test = range(2, 8)
silhouette_scores = []

print("Silhouette Analysis Results:")
print(f"{'K':>5} | {'Mean Silhouette Score':>22}")
print("-" * 50)

# Compute silhouette scores
for k in K_test:
    km = KMeans(n_clusters=k, init='k-means++',
                n_init=10, random_state=42, max_iter=300)
    
    labels = km.fit_predict(df_scaled)
    score = silhouette_score(df_scaled, labels)
    
    silhouette_scores.append(score)
    
    print(f"{k:>5} | {score:>22.4f}")

print("-" * 50)

# Identify best K
best_k = K_test[silhouette_scores.index(max(silhouette_scores))]
best_score = max(silhouette_scores)

# Plot
plt.figure(figsize=(8, 5))
plt.plot(K_test, silhouette_scores, marker='o', linewidth=2)

# Highlight best K
plt.axvline(x=best_k, linestyle='--', label=f'Best K = {best_k}')

plt.xlabel('Number of Clusters (K)')
plt.ylabel('Mean Silhouette Score')
plt.title('Silhouette Score vs Number of Clusters')
plt.xticks(list(K_test))
plt.legend()
plt.grid(alpha=0.3)
plt.show()

print(f"\nBest K based on silhouette score: {best_k} (Score = {best_score:.4f})")
```

**Insight:**
- Looking at the results of the silhouette analysis, it can be seen that K = 2 had the highest silhouette score which could mean that it is a well seperated cluster compared to the other Ks.
- Seeing that both Elbow Method and Silhouette analysis show that K = 2 as the best K for the given dataset. Using this K is the best since it is the most compact and well-seperated cluster. This would ensure that the group would get the best results.

## **Part 3 — DBSCAN**

**DBSCAN (Density-Based Spatial Clustering of Applications with Noise)** is a clustering algorithm that groups data points based on density rather than distance to a centroid. It identifies clusters as regions of high point concentration separated by areas of low density, while explicitly labeling sparse points as noise or outliers. This makes DBSCAN particularly suitable for real-world datasets with irregular cluster shapes and the presence of anomalies. In this study, DBSCAN is applied to selected credit card usage features to uncover natural customer segments and detect atypical financial behavior.

### **Part 3A — k-distance Plot**

```python
X_scaled = df_scaled.values

# Use min_samples = 5
k = 5

neighbors = NearestNeighbors(n_neighbors=k)
neighbors_fit = neighbors.fit(X_scaled)
distances, indices = neighbors_fit.kneighbors(X_scaled)

# Sort distances to the 5th nearest neighbor
k_distances = np.sort(distances[:, k - 1])
```

```python
# Plot k-distance graph
plt.figure(figsize=(8,5))
plt.plot(k_distances)
plt.title("k-distance Plot (k = 5)")
plt.xlabel("Points sorted by distance")
plt.ylabel("Distance to 5th nearest neighbor")
plt.grid(True)
plt.show()
```

**Insight:**
- The k-distance plot shows a clear elbow around eps ≈ 1.0, where the curve transitions from a gradual increase to a sharp rise. This point marks the boundary between dense regions (clusters) and sparse regions (noise).
- Points before the elbow belong to well-defined clusters, while points after the elbow represent outliers or low-density areas.
- The presence of a distinct elbow indicates a suitable density threshold, supporting the choice of eps = 1.0 for effective cluster formation while limiting noise.

### **Part 3B — Test Multiple eps Values**

```python
eps_values = [0.5, 0.8, 1.0, 1.2]

results = []

for eps in eps_values:
    dbscan = DBSCAN(eps=eps, min_samples=5)
    labels = dbscan.fit_predict(X_scaled)

    n_clusters = len(set(labels)) - (1 if -1 in labels else 0)
    noise_points = np.sum(labels == -1)
    noise_percent = (noise_points / len(labels)) * 100

    results.append([eps, n_clusters, noise_points, noise_percent])

eps_results = pd.DataFrame(results, columns=["eps", "clusters_found", "noise_points", "noise_percent"])
eps_results
```

**Interpretation:**  

The k-distance plot shows an elbow at **eps ≈ 1.0**, indicating a suitable clustering threshold.

At **eps = 0.5**, the model over-segments the data (9 clusters, 13.96% noise). As eps increases to **0.8, 1.0, and 1.2**, clusters stabilize at 4 while noise decreases, showing improved structure capture.

Although **eps = 1.2** has the lowest noise (2.56%), it risks merging clusters. Thus, **eps = 1.0** is optimal, balancing stable clusters (4) and moderate noise (3.58%) without over-merging.

**Final parameters:**
* eps = 1.0
* min_samples = 5

### **Part 3C — Final Parameter Selection**

```python
# Final chosen parameters
final_eps = 1.0
final_min_samples = 5

dbscan_final = DBSCAN(eps=final_eps, min_samples=final_min_samples)
labels = dbscan_final.fit_predict(X_scaled)

# Add cluster labels to the cleaned dataset
df_dbscan = df_select.copy()
df_dbscan["Cluster"] = labels

df_dbscan.head()
```

This can easily help marketing teams, as it can directly support targeted strategies such as budget-friendly promotions versus premium offers.

In contrast, DBSCAN produces highly imbalanced clustering results, where the majority of customers are assigned to a single dominant cluster, while smaller clusters contain extremely few observations. This suggests that DBSCAN is detecting only a few dense clusters while classifying most customers as part of a broad general behavior group. Additionally, DBSCAN identifies noise points, which may represent a random or extreme behavioral patterns, including both unusually high and low activity customers. However, these noise points are too sparse and random to form a simple marketing solution.

From a metrics perspective, DBSCAN demonstrates its strengths in cluster separation, achieving a higher Silhouette Score (0.6076) and a much lower Davies–Bouldin Index (0.32), indicating well-separated and coherent clusters. However, K-Means shows a significantly higher Calinski–Harabasz Index (3111.39), which suggests stronger overall cluster compactness and clearer structure in the dataset. This is important because marketing segmentation benefits more from stable and consistent groupings rather than highly sparse and random structures.

While DBSCAN is effective in identifying anomalies or rare customer behaviors, these noise points are not easily actionable for a marketing team. In contrast, K-Means provides a simpler and business-friendly model that is easier to interpret, communicate, and deploy.

Overall, despite DBSCAN’s stronger separation metrics, K-Means is more suitable for practical business use due to its interpretability, stability, and actionable cluster structure.
