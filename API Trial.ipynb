```python
import requests
import pandas as pd

# Try access PHP historical data from Frankfurter API
def check_api():
    # Fetcl data last 30 days for USD/PHP and JPY/PHP
    url = "https://api.frankfurter.app/latest?from=PHP&to=USD,JPY"
    
    try:
        response = requests.get(url)
        data = response.json()
        print(f"Connection Successful! Base: {data['base']}")
        print(f"Date: {data['date']}")
        print(f"Rates: {data['rates']}")
    except Exception as e:
        print(f"Connection Failed: {e}")

check_api()
```

```python
# Try: Accessing and displaying specific currency pairs
def check_api():
    # Fetch latest rates with PHP as its base currency
    url = "https://api.frankfurter.app/latest?from=PHP&to=USD,JPY"
    
    try:
        response = requests.get(url)
        # To raise an exception if the request was unsuccessful
        response.raise_for_status() 
        
        data = response.json()
        
        # Extracting data from the JSON response
        date = data.get('date')
        rates = data.get('rates', {})
        
        # Specific conversions
        php_to_usd = rates.get('USD')
        php_to_jpy = rates.get('JPY')
        
        print(f"--- Exchange Rates as of {date} ---")
        print(f"Base Currency: 1 PHP")
        print(f"US Dollar (USD): {php_to_usd:.4f}")
        print(f"Japanese Yen (JPY): {php_to_jpy:.2f}")
        print("-" * 10)
        print("Connection Successful!")

    except Exception as e:
        print(f"Connection Failed: {e}")

check_api()
```

```python
# Try: Accessing Foreign Currency to PHP Conversion
def check_api():
    try:
        # Call 1: Get USD to PHP
        usd_url = "https://api.frankfurter.app/latest?from=USD&to=PHP"
        usd_res = requests.get(usd_url)
        usd_res.raise_for_status()
        usd_to_php = usd_res.json()['rates']['PHP']

        # Call 2: Get JPY to PHP
        jpy_url = "https://api.frankfurter.app/latest?from=JPY&to=PHP"
        jpy_res = requests.get(jpy_url)
        jpy_res.raise_for_status()
        jpy_to_php = jpy_res.json()['rates']['PHP']
        
        date = usd_res.json()['date']

        print(f"--- PHP Exchange Rates as of {date} ---")
        print(f"1 US Dollar (USD)    = ₱{usd_to_php:.2f} PHP")
        print(f"1 Japanese Yen (JPY)  = ₱{jpy_to_php:.4f} PHP")
        print("-" * 20)
        print("Connection Successful!")

    except Exception as e:
        print(f"Connection Failed.")
        print(f"Error: {e}")

check_api()
```
